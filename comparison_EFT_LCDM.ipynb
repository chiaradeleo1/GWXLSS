```python
import numpy as np
import matplotlib.pyplot as plt

import seaborn           as sb
import pandas as pd
sidelegend = {'bbox_to_anchor': (1.04,0.5), 
              'loc': "center left",
              'frameon': False}

filename_old = './mock_data/MG_test_galonly_IApoly_Cls_noiseless.dat'
#filename_old= './mock_data/LCDM_test_galonly_sources_Cls_noiseless.dat'
data_old = data = pd.read_csv(filename_old, delim_whitespace=True)
filename= './mock_data/LCDM_test_galonly_Cls_noiseless.dat'
#filename= './mock_data/LCDM_N=10^5_Euclid_Cls.dat'
data = pd.read_csv(filename, delim_whitespace=True)

ells = data['ells'].values

bincolors = sb.color_palette('hls',11)
norm = ells * (ells + 1) / (2 * np.pi)
Nbins_gc=10
Nbins_wl=10
use_obs  = ['GC','WL']



fig, axes = plt.subplots(ncols=3, sharey=False, subplot_kw=dict(frameon=True),figsize=(12,4))
plt.suptitle('Comparison $C_\ell$ with LCDM (o) and MG (-) EFTcode ',y=1.05)

if 'GC' in use_obs:
    axes[0].set_title('GCph',loc='right',fontdict={'fontsize':16})
    for i in range(1,Nbins_gc+1):
        #print(data_old['G{}xG{}'.format(i,i)]-data['G{}xG{}'.format(i,i)])
        axes[0].plot(ells,data_old['G{}xG{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1])
        axes[0].plot(ells,data['G{}xG{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1],  marker='o')
    axes[0].set_xlabel(r'$\ell$')
    axes[0].ticklabel_format(axis='y', style='sci', scilimits=(0,0))
    axes[0].set_xscale('log')
    axes[0].set_yscale('log')
    
if 'WL' in use_obs:
    axes[1].set_title('WL',loc='right',fontdict={'fontsize':16})
    for i in range(1,Nbins_wl+1):
        # print(data_old['L{}xL{}'.format(i,i)]-data['L{}xL{}'.format(i,i)])
         axes[1].plot(ells,data_old['L{}xL{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1])
         axes[1].plot(ells,data['L{}xL{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1],  marker='o')
    axes[1].set_xlabel(r'$\ell$')
    axes[1].ticklabel_format(axis='y', style='sci', scilimits=(0,0))
    axes[1].set_xscale('log')
    axes[1].set_yscale('log')
    
if 'GC' in use_obs and 'WL' in use_obs:
    axes[2].set_title('GGL',loc='right',fontdict={'fontsize':16})
    for i in range(1,Nbins_gc+1):
         axes[2].plot(ells,data_old['G{}xL{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1])
         axes[2].plot(ells,data['G{}xL{}'.format(i,i)]*norm,
                     label=r'$i={}$'.format(i),color=bincolors[i-1], marker='o', ls=':')
    axes[2].set_xlabel(r'$\ell$')
    axes[2].ticklabel_format(axis='y', style='sci', scilimits=(0,0))
    axes[2].set_xscale('log')

axes[0].set_ylabel(r'$C_\ell^{ixi}$')
plt.subplots_adjust(hspace=1.2)
axes[-1].legend(**sidelegend)
#plt.savefig('Cls_comparison_MG.png', dpi=300, bbox_inches='tight')
plt.show
```
